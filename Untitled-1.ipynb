```python
# Step 1: Importing Required Libraries
# =====================================
# For data manipulation and numerical operations
import pandas as pd
import numpy as np

# For data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# For preprocessing and modeling
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    classification_report, 
    confusion_matrix, 
    roc_auc_score, 
    precision_recall_curve, 
    auc,
    precision_score, 
    recall_score, 
    f1_score, 
    average_precision_score
)

# For handling class imbalance
from imblearn.over_sampling import SMOTE

# For building the hybrid model
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras import layers, losses
import xgboost as xgb
from sklearn.linear_model import LogisticRegression

print("Libraries imported successfully.")
```

```python
# --- Loading the PRE-SPLIT data ---

try:
    X_train = pd.read_csv('X_train_raw.csv')
    X_test = pd.read_csv('X_test_raw.csv')
    
    y_train = pd.read_csv('y_train_raw.csv').squeeze()
    y_test = pd.read_csv('y_test_raw.csv').squeeze()
    
    print("Pre-split training and testing data loaded successfully.")
    print("Training set shape:", X_train.shape)
    print("Testing set shape:", X_test.shape)
except FileNotFoundError:
    print("Error: Pre-split files not found. Please run the 'Run-Once' data splitting script first.")
```

```python
# --- Feature Engineering Function ---
def engineer_features(dataframe):
    df_eng = dataframe.copy()
    # Converting time columns to datetime objects
    df_eng['trans_date_trans_time'] = pd.to_datetime(df_eng['trans_date_trans_time'])
    df_eng['dob'] = pd.to_datetime(df_eng['dob'])
    
    # Creating time-based features
    df_eng['trans_hour'] = df_eng['trans_date_trans_time'].dt.hour
    df_eng['trans_day_of_week'] = df_eng['trans_date_trans_time'].dt.dayofweek
    
    # Creating age feature
    df_eng['age'] = (df_eng['trans_date_trans_time'] - df_eng['dob']).dt.days // 365
    
    # Creating distance feature
    df_eng['distance'] = np.sqrt((df_eng['merch_lat'] - df_eng['lat'])**2 + (df_eng['merch_long'] - df_eng['long'])**2)
    
    return df_eng

# Apply feature engineering to both train and test sets
X_train_eng = engineer_features(X_train)
X_test_eng = engineer_features(X_test)

# --- Defining Columns for Preprocessing ---
# Droping original and identifier columns that are no longer needed
drop_cols = ['Unnamed: 0', 'cc_num', 'first', 'last', 'street', 'city', 'state', 'zip', 'trans_num', 'merchant', 'trans_date_trans_time', 'dob', 'lat', 'long', 'merch_lat', 'merch_long']

categorical_features = ['category', 'gender', 'job']
numerical_features = ['amt', 'city_pop', 'unix_time', 'trans_hour', 'trans_day_of_week', 'age', 'distance']

# --- Creating Preprocessing Pipeline ---
# This pipeline will scale numerical features and one-hot encode categorical features

# Creating the individual transformers
numerical_transformer = StandardScaler()
categorical_transformer = OneHotEncoder(handle_unknown='ignore')

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ],
    remainder='drop' # Drop columns not specified in the transformers
)

# --- Applying Preprocessing ---

X_train_processed = preprocessor.fit_transform(X_train_eng)
X_test_processed = preprocessor.transform(X_test_eng)

print("Feature engineering and preprocessing complete.")
print("Shape of processed training data:", X_train_processed.shape)
```

```python

print("[X_train_processed, y_train] (Unbalanced training data)")
print("[X_test_processed, y_test] (Test data)\n")

# --- Case 1: Logistic Regression on UNBALANCED Data ---
print("Case 1: Training on UNBALANCED Data (with class_weight='balanced')\n")

# We use class_weight='balanced' to help the model handle the imbalanced data
logreg_unbalanced = LogisticRegression(class_weight='balanced', random_state=42, max_iter=1000)

try:
    # Train the model
    logreg_unbalanced.fit(X_train_processed, y_train)
    print("Model trained successfully on unbalanced data.")

    # --- Evaluation on Test Set ---
    # Getting class predictions (0 or 1)
    y_pred_unbalanced = logreg_unbalanced.predict(X_test_processed)
    
    # Getting prediction probabilities for the positive class (fraud)
    #for AUPRC
    y_probs_unbalanced = logreg_unbalanced.predict_proba(X_test_processed)[:, 1]

    # Calculate metrics
    precision_unbalanced = precision_score(y_test, y_pred_unbalanced)
    recall_unbalanced = recall_score(y_test, y_pred_unbalanced)
    f1_unbalanced = f1_score(y_test, y_pred_unbalanced)
    
    # Calculate AUPRC (Area Under the Precision-Recall Curve)
    auprc_unbalanced = average_precision_score(y_test, y_probs_unbalanced)

    print("\n--- Evaluation Metrics (Unbalanced Model on Test Data) ---")
    print(f"Precision: {precision_unbalanced:.4f}")
    print(f"Recall:    {recall_unbalanced:.4f}")
    print(f"F1-Score:  {f1_unbalanced:.4f}")
    print(f"AUPRC:     {auprc_unbalanced:.4f}")
    
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred_unbalanced))

except NameError as e:
    print(f"\nError: A required variable is not defined. Make sure all data is loaded.")
    print(f"Details: {e}")
except Exception as e:
    print(f"\nAn unexpected error occurred during unbalanced model training/evaluation: {e}")

print("\nUnbalanced Logistic Regression evaluation complete.")
```

```python
# --- Isolating Legitimate Transactions for Training ---
# Converting the sparse matrix to a dense numpy array for boolean indexing
X_train_processed_dense = X_train_processed.toarray()
X_train_legit = X_train_processed_dense[y_train == 0]

# --- Defining the Autoencoder Architecture ---
input_dim = X_train_processed.shape[1]
encoding_dim = 32 # hyperparameter - tuning possible

input_layer = layers.Input(shape=(input_dim,))
encoder = layers.Dense(encoding_dim, activation="relu")(input_layer)
decoder = layers.Dense(input_dim, activation='sigmoid')(encoder)
autoencoder = Model(inputs=input_layer, outputs=decoder)

autoencoder.compile(optimizer='adam', loss='mean_squared_error')
autoencoder.summary()

# --- Training the Autoencoder ---
print("\nTraining Autoencoder on legitimate transactions only...")
history = autoencoder.fit(X_train_legit, X_train_legit,
                          epochs=10,
                          batch_size=256,
                          shuffle=True,
                          validation_split=0.2,
                          verbose=1)

print("\nAutoencoder training complete.")
```

```python
# --- Calculating Reconstruction Error in Batches to Avoid Memory Errors ---

batch_size = 3500 

# --- Processing Training Data in Batches ---
print("Calculating reconstruction error for training data (in batches)...")
# Initializing an empty list to store the loss values from each batch
train_loss_list = list()
# Looping through the training data in steps of batch_size
for i in range(0, X_train_processed_dense.shape[0], batch_size):
    batch = X_train_processed_dense[i:i+batch_size]
    # Predicting on the small batch
    reconstructions_batch = autoencoder.predict(batch, verbose=0)
    # Calculating the loss for this small batch immediately
    batch_loss = tf.keras.losses.mae(reconstructions_batch, batch)
    # Appending the resulting loss values to our list
    train_loss_list.append(batch_loss)

# --- Combining the results and create the new feature ---

train_loss = np.concatenate(train_loss_list).astype(np.float32)


X_train_processed_dense_f32 = X_train_processed_dense.astype(np.float32)


X_train_with_error = np.c_[X_train_processed_dense_f32, train_loss.reshape(-1, 1)]


del X_train_processed_dense
del X_train_processed_dense_f32
del train_loss


# --- Processing Test Data in Batches ---
print("Calculating reconstruction error for test data (in batches)...")
X_test_processed_dense = X_test_processed.toarray()
test_loss_list = list()
for i in range(0, X_test_processed_dense.shape[0], batch_size):
    batch = X_test_processed_dense[i:i+batch_size]
    reconstructions_batch = autoencoder.predict(batch, verbose=0)
    batch_loss = tf.keras.losses.mae(reconstructions_batch, batch)
    test_loss_list.append(batch_loss)

# --- Combining the results and create the new feature ---

test_loss = np.concatenate(test_loss_list).astype(np.float32)


X_test_processed_dense_f32 = X_test_processed_dense.astype(np.float32)

X_test_with_error = np.c_[X_test_processed_dense_f32, test_loss.reshape(-1, 1)]


del X_test_processed_dense
del X_test_processed_dense_f32
del test_loss

print("\nReconstruction error feature created and added to train and test sets.")
print("New shape of training data:", X_train_with_error.shape)
```

```python

from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler

# --- Applying the Hybrid Sampling Strategy ---
print("Original training distribution:\n", y_train.value_counts())


X_train_with_error_f32 = X_train_with_error.astype(np.float32)

# --- Defining the Hybrid Strategy ---
# Step 1: Oversample the minority class (fraud) to be 10% of the majority class
over = SMOTE(sampling_strategy=0.1, random_state=42)
# Step 2: Undersample the majority class (legitimate) to be 50% of the new minority class size
# This will result in a 2:1 ratio of legitimate to fraud samples.
under = RandomUnderSampler(sampling_strategy=0.5, random_state=42)

#Creating a pipeline to execute the steps in order
pipeline = Pipeline(steps=[('o', over), ('u', under)])


print("\nApplying Hybrid Sampling (SMOTE + Undersampling). This may take a moment...")
X_train_resampled, y_train_resampled = pipeline.fit_resample(X_train_processed, y_train)
print("Resampling complete.")
print("\nResampled training distribution:\n", pd.Series(y_train_resampled).value_counts())


del X_train_with_error_f32

#Defining and Train the Logistic Regression on the hybrid sampled data
logreg_balanced = LogisticRegression()
logreg_balanced.fit(X_train_resampled, y_train_resampled)


# --- Defining and Train the XGBoost Classifier ---
xgb_classifier = xgb.XGBClassifier(objective='binary:logistic',
                                   eval_metric='logloss',
                                   use_label_encoder=False,
                                   n_estimators=200,
                                   learning_rate=0.1,
                                   max_depth=5,
                                   random_state=42)

print("\nTraining XGBoost classifier on the hybrid-sampled data...")
xgb_classifier.fit(X_train_resampled, y_train_resampled)
print("XGBoost training complete.")
```

```python
# --- Case 2: Logistic Regression on BALANCED (Resampled) Data ---
print("Case 2: Training on BALANCED (Resampled) Data\n")

logreg_balanced = LogisticRegression(random_state=42, max_iter=1000)

try:
    # Train the model
    logreg_balanced.fit(X_train_resampled, y_train_resampled)
    print("Model trained successfully on balanced (resampled) data.")

    # --- Evaluation on Test Set ---
    # Get class predictions (0 or 1)
    y_pred_balanced = logreg_balanced.predict(X_test_processed)
    
    # Get prediction probabilities for the positive class (fraud)
    y_probs_balanced = logreg_balanced.predict_proba(X_test_processed)[:, 1]

    # Calculate metrics
    precision_balanced = precision_score(y_test, y_pred_balanced)
    recall_balanced = recall_score(y_test, y_pred_balanced)
    f1_balanced = f1_score(y_test, y_pred_balanced)
    
    # Calculate AUPRC
    auprc_balanced = average_precision_score(y_test, y_probs_balanced)

    print("\n--- Evaluation Metrics (Balanced Model on Test Data) ---")
    print(f"Precision: {precision_balanced:.4f}")
    print(f"Recall:    {recall_balanced:.4f}")
    print(f"F1-Score:  {f1_balanced:.4f}")
    print(f"AUPRC:     {auprc_balanced:.4f}")

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred_balanced))

except NameError as e:
    print(f"\nError: A required variable is not defined. Make sure all data is loaded.")
    print(f"Details: {e}")
except Exception as e:
    print(f"\nAn unexpected error occurred during balanced model training/evaluation: {e}")

print("\nBalanced Logistic Regression evaluation complete.")
```

```python
print("--- Analyzing Individual Predictions (Testing Phase) ---")

def explain_transaction(index_position):
    print(f"\n--- Explaining Transaction at Test Set Position: {index_position} ---")
    
    # ✅ Handling both NumPy and Pandas
    if hasattr(X_test_with_error, "iloc"):
        single_transaction = X_test_with_error.iloc[index_position].values
    else:
        single_transaction = X_test_with_error[index_position]
    
    # ✅ Handling y_test
    true_label = y_test.iloc[index_position] if hasattr(y_test, "iloc") else y_test[index_position]
    
    # ✅ Making prediction
    prediction = xgb_classifier.predict(single_transaction.reshape(1, -1))
    
    print(f"True Label: {'Fraud' if true_label == 1 else 'Legitimate'}")
    print(f"Model Prediction: {'Fraud' if prediction == 1 else 'Legitimate'}")
    
    # ✅ Generate SHAP 
    shap_values_single = explainer.shap_values(single_transaction.reshape(1, -1))
    
    shap.waterfall_plot(
        shap.Explanation(
            values=shap_values_single[0],
            base_values=explainer.expected_value,
            data=single_transaction,
            feature_names=all_feature_names
        ),
        show=True
    )




fraud_ilocs_array = np.where(y_test == 1)[0]
if fraud_ilocs_array.size > 0:
    first_fraud_iloc = fraud_ilocs_array[0]
    explain_transaction(first_fraud_iloc)
else:
    print("\nNo fraudulent transactions found in the test set to explain.")

legit_ilocs_array = np.where(y_test == 0)[0]
if legit_ilocs_array.size > 0:
    first_legit_iloc = legit_ilocs_array[0]
    explain_transaction(first_legit_iloc)
else:
    print("\nNo legitimate transactions found in the test set to explain.")
```
